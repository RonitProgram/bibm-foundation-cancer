```python
import pandas as pd
import os

base_dir = "/lustre/nvwulf/home/ronsharma"
file_path = "pone.0126631.s001.xlsx"  # replace with your downloaded filename

excel_file = pd.ExcelFile(os.path.join(base_dir, file_path))

print("Sheet names:")
print(excel_file.sheet_names)
```

```python
for sheet in excel_file.sheet_names:
    print(f"\n--- {sheet} ---")
    preview = pd.read_excel(
        file_path,
        sheet_name=sheet,
        header=None,
        nrows=10
    )
    print(preview.to_string(index=False, header=False))
```

```python
msig = pd.read_excel(os.path.join(base_dir, file_path), sheet_name="Table B in S1 File")

print(msig.shape)
print(msig.columns.tolist())
print(msig.head())
```
